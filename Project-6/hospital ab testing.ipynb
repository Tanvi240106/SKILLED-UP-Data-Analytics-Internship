```python
import pandas as pd

# Load all tables
patients = pd.read_csv("C:/Users/tanvi jalan/Downloads/patients csv.csv")
admissions = pd.read_csv("C:/Users/tanvi jalan/Downloads/admissions csv.csv")
departments = pd.read_csv("C:/Users/tanvi jalan/Downloads/departments csv.csv")
treatments = pd.read_csv("C:/Users/tanvi jalan/Downloads/treatments csv.csv")
billing = pd.read_csv("C:/Users/tanvi jalan/Downloads/billing csv.csv")
readmissions = pd.read_csv("C:/Users/tanvi jalan/Downloads/readmissions csv.csv")

```

```python
# Merge in logical order
df = admissions.merge(patients, on='patient_id', how='left') \
               .merge(treatments, on='admission_id', how='left') \
               .merge(billing, on='admission_id', how='left') \
               .merge(readmissions, on=['patient_id', 'admission_id'], how='left') \
               .merge(departments, on='department', how='left')
```

```python
# Ensure column exists and is binary
df['was_readmitted'] = df['readmit_within_30_days'].apply(lambda x: 1 if str(x).lower() == 'yes' else 0)
```

```python
df['readmit_within_30_days'] = df['readmit_within_30_days'].fillna('no')
df['was_readmitted'] = df['readmit_within_30_days'].apply(lambda x: 1 if str(x).strip().lower() == 'yes' else 0)
```

```python
group_angioplasty = df[df['treatment_type'] == 'Angioplasty']['was_readmitted']
group_bypass = df[df['treatment_type'] == 'Bypass']['was_readmitted']
```

```python
from scipy.stats import ttest_ind

# Mean readmission rate
print("Angioplasty Readmission Rate:", round(group_angioplasty.mean(), 3))
print("Bypass Readmission Rate:", round(group_bypass.mean(), 3))

# Perform t-test
t_stat, p_value = ttest_ind(group_angioplasty, group_bypass)

print("T-statistic:", round(t_stat, 4))
print("P-value:", round(p_value, 4))

# Interpret
if p_value < 0.05:
    print("✅ Statistically significant difference!")
    better = "Bypass" if group_angioplasty.mean() > group_bypass.mean() else "Angioplasty"
    print(f"{better} is better (lower readmission rate).")
else:
    print("❌ No significant difference found.")
```
